###2. task:

Create a House class that represents a house; attributes:


*   area
*   number of rooms
*   number of bathrooms
*   number of floors
*   has garage
*   has garden
*   any other attributes that you think is important



instance methods:

*   price calculation

There should be at least 1 default parameter

(4 points)

```python
class House:
    def __init__(self, area, num_rooms, num_bathrooms, num_floors,
                 has_garage=False, has_garden=False, garage_spaces=0):
        self.area = area
        self.num_rooms = num_rooms
        self.num_bathrooms = num_bathrooms
        self.num_floors = num_floors
        self.has_garage = has_garage
        self.has_garden = has_garden
        self.garage_spaces = garage_spaces

    def calculate_price(self):
        price = self.area * 1000

        if self.has_garage:
            price += 15000

        if self.has_garden:
            price += 10000

        price += (self.num_floors - 1) * 20000

        return price
```

Create a Person class with the following attributes:


*   name
*   age
*   married (T/F)
*   number of children
*   number of cars
*   total amount of money (to buy a house)
*   based on the number of children store the number of family members in an attribute  (- this does not have to be included in the init arguments as it is a calculated value)


(4 points)




```python
class Person:
    def __init__(self, name, age, married, num_children, num_cars, total_money):
        self.name = name
        self.age = age
        self.married = married
        self.num_children = num_children
        self.num_cars = num_cars
        self.total_money = total_money
        self.family_members = self.calculate_family_members()

    def calculate_family_members(self):
        if self.married:
            return 2 + self.num_children
        else:
            return 1 + self.num_children
```

```python
# FUNCTIONS

def can_buy_house(person, house):
    return person.total_money >= house.calculate_price()

def enough_rooms(person, house):
    return house.num_rooms * 2 >= person.family_members

def enough_garage_space(person, house):
    return house.garage_spaces >= person.num_cars
```

###A) (5 points)
 Create an Animal class and choose 3 animals that have some things in common (characteristics and behaviour)!

Write the Parent class and the derived classes of the species!

The Child classes should inherit some attributes and override some methods of the Parent class, but should also extend the state with own attributes in each subclass and extend the behaviour with own methods.

Decide yourself:

which methods belong to parent


which should be overridden


which attributes are shared. Explain your choice.

```python

class Animal:
    def __init__(self, name, age, habitat):
        self.name = name
        self.age = age
        self.habitat = habitat

    # shared behavior
    def eat(self):
        return f"{self.name} is eating."

     # overridden
    def sound(self):
        return f"{self.name} makes a sound."
     # shared
    def move(self):
        return f"{self.name} is moving."

# DOG
class Dog(Animal):
    def __init__(self, name, age, habitat, breed, trained):
        super().__init__(name, age, habitat)
        self.breed = breed
        self.trained = trained

    def sound(self):
        return f"{self.name} says: Woof!"

    def fetch(self):
        return f"{self.name} is fetching the ball."

# CAT
class Cat(Animal):
    def __init__(self, name, age, habitat, color, indoor):
        super().__init__(name, age, habitat)
        self.color = color
        self.indoor = indoor

    def sound(self):
        return f"{self.name} says: Meow!"

    def scratch(self):
        return f"{self.name} is scratching the sofa."

# BIRD
class Bird(Animal):
    def __init__(self, name, age, habitat, wing_span, can_fly):
        super().__init__(name, age, habitat)
        self.wing_span = wing_span
        self.can_fly = can_fly

    def sound(self):
        return f"{self.name} says: Chirp!"

    def fly(self):
        if self.can_fly:
            return f"{self.name} is flying!"
        else:
            return f"{self.name} cannot fly."
```

Create classes and use inheritance to calculate the area of different shapes (specifically, circle and square, you can add triangle or anything else if you want).

Requirements:


1.	Create a base class called Shape that has a method named area().

This method should not have any implementation
2. Create at least two subclasses:

Circle — which takes a radius as a parameter

Square — which takes a side as a parameter

write the area-calculating function for both (or for all)

3. Create objects of both Circle and Square, and print their areas.

Extra: Store these objects in a list and use a loop to print the area of each shape.

```python
# BASE CLASS
class Shape:
    def area(self):
        # No implementation (as required)
        pass

# CIRCLE CLASS
class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.1416 * self.radius * self.radius


#SQUARE CLASS
class Square(Shape):
    def __init__(self, side):
        self.side = side

    def area(self):
        return self.side * self.side

# TRIANGLE CLASS
class Triangle(Shape):
    def __init__(self, base, height):
        self.base = base
        self.height = height

    def area(self):
        return 0.5 * self.base * self.height
```

```python
circle1 = Circle(5)
square1 = Square(4)
triangle1 = Triangle(6, 3)

print("Circle area:", circle1.area())
print("Square area:", square1.area())
print("Triangle area:", triangle1.area())
```

Count character frequency

```python
text = "banana"
```

Expected output: {'b': 1, 'a': 3, 'n': 2}

```python
freq = {}
for char in text:
    if char in freq:
        freq[char] += 1
    else:
        freq[char] = 1

print(freq)
```
